```python
# DATA CLEANING PLAN
# ==================

# Fixing Order_Date (30 missing - 0.4%)?
# Delete them
# Reason: We don't know when the order was placed, it will be an issue during analysis

# Fixing Price null values(37 missing - 0.5%)?
# DROP these rows
# Reason: Without price we don't know how much the customer spent - this is critical

# Fixing Quantity null values(35 missing - 0.4%)?
# DROP these rows
# Reason: Quantity tells us if customer is wholesale or individual buyer
# It's only 0.4% of data so removing it won't hurt

# Fixing Payment_Method null values(1672 missing - 20%)?
# Replace with "Unknown"
# Reason: User needs to pay if the order was finalized, the method was not registered.
# FOR THIS PROJECT I WILL ASSUME THAT NaN VALUES ARE THE SYSTEM ERROR.

# Fixing Discount_Percent null values(1203 missing - 15%)?
# Replace with 0
# Reason:I assume No discount = 0% discount

# Fixing Duplicates (18 rows)?
# Delete them
# Reason: Impossible that someone ordered exactly same items at same second
```

```python
df = df.dropna(subset =['Order_Date'])
df = df.dropna(subset =['Price'])
df = df.dropna(subset =['Quantity'])
df['Payment_Method'] = df['Payment_Method'].fillna('Unknown')
df['Discount_Percent'] = df['Discount_Percent'].fillna(0)
df = df.drop_duplicates(subset = ['Order_ID'])
```
